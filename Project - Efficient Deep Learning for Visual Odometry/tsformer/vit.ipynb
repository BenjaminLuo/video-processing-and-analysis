```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# 7DOF Transformation Functions
def scale_lse_solver(X, Y):
    """Least-sqaure-error solver for optimal scaling factor."""
    scale = np.sum(X * Y) / np.sum(X ** 2)
    return scale

def umeyama_alignment(x, y, with_scale=False):
    """Umeyama alignment for 7DOF transformation."""
    if x.shape != y.shape:
        raise ValueError("x.shape not equal to y.shape")

    m, n = x.shape
    mean_x = x.mean(axis=1)
    mean_y = y.mean(axis=1)
    sigma_x = 1.0 / n * (np.linalg.norm(x - mean_x[:, np.newaxis])**2)
    cov_xy = (1.0 / n) * (y - mean_y[:, np.newaxis]) @ (x - mean_x[:, np.newaxis]).T

    U, d, Vt = np.linalg.svd(cov_xy)
    S = np.eye(m)
    if np.linalg.det(U) * np.linalg.det(Vt.T) < 0:
        S[-1, -1] = -1

    R = U @ S @ Vt
    c = np.trace(S @ np.diag(d)) / sigma_x if with_scale else 1.0
    t = mean_y - c * R @ mean_x

    return R, t, c

# Load trajectory data
ground_truth_path = "/Users/epheriami/Downloads/4B/Code/SYDE 673/P/data/poses/10.txt"
prediction_path = "/Users/epheriami/Downloads/4B/Code/SYDE 673/P/tsformer/checkpoints/Exp3/checkpoint_model3_exp20/pred_poses/10.txt"

with open(ground_truth_path, "r") as f:
    ground_truth_data = f.read()

with open(prediction_path, "r") as f:
    prediction_data = f.read()

def parse_trajectory(data):
    """Parse trajectory data including X, Y, Z translations."""
    trajectory = []
    lines = data.strip().split('\n')
    for line in lines:
        parts = list(map(float, line.strip().split()))
        if len(parts) >= 12:
            tx, ty, tz = parts[3], parts[7], parts[11]
            trajectory.append((tx, ty, tz))
    return trajectory

# Parse trajectories
gt_trajectory = parse_trajectory(ground_truth_data)
pred_trajectory = parse_trajectory(prediction_data)

# Configure image path
sequence_path = "/Users/epheriami/Downloads/4B/Code/SYDE 673/P/data/sequences_jpg/10"
image_folder = os.path.join(sequence_path, "image_0")
image_files = sorted([os.path.join(image_folder, f) 
                     for f in os.listdir(image_folder) if f.endswith(".jpg")])

# Verify we have enough frames
n_frames = min(len(gt_trajectory), len(pred_trajectory), len(image_files))
gt_trajectory = gt_trajectory[:n_frames]
pred_trajectory = pred_trajectory[:n_frames]

# Apply 7DOF transformation to align predicted trajectory
gt_xyz = np.array(gt_trajectory).T
pred_xyz = np.array(pred_trajectory).T
R, t, scale = umeyama_alignment(pred_xyz, gt_xyz, with_scale=True)
transformed_pred = (scale * (R @ pred_xyz) + t.reshape(3, 1)).T

# Extract transformed coordinates for plotting
gt_x, gt_z = [p[0] for p in gt_trajectory], [p[2] for p in gt_trajectory]
pred_x, pred_z = transformed_pred[:, 0], transformed_pred[:, 2]

# Pre-load all images
images = [cv2.cvtColor(cv2.imread(f), cv2.COLOR_BGR2RGB) for f in image_files[:n_frames]]

# Set up visualization
fig, (ax_video, ax_traj) = plt.subplots(1, 2, figsize=(14, 6))

# Video display setup
ax_video.set_title("KITTI Video Sequence")
ax_video.axis('off')
video_display = ax_video.imshow(images[0])

# Trajectory plot setup
ax_traj.set_title("Visual Odometry Trajectory (7DOF Aligned)")
ax_traj.set_xlabel("X Position (m)")
ax_traj.set_ylabel("Z Position (m)")
ax_traj.grid(True)
ax_traj.axis('equal')

# Initialize trajectory lines
gt_line, = ax_traj.plot([], [], 'b-', label='Ground Truth', linewidth=2)
pred_line, = ax_traj.plot([], [], 'r--', label='Prediction', linewidth=2)
ax_traj.legend()

def update(frame):
    """Update function for animation."""
    video_display.set_array(images[frame])
    ax_video.set_title(f"Frame {frame+1}/{n_frames}")
    
    gt_line.set_data(gt_x[:frame+1], gt_z[:frame+1])
    pred_line.set_data(pred_x[:frame+1], pred_z[:frame+1])
    
    # Adjust view limits dynamically
    all_x = gt_x[:frame+1] + pred_x[:frame+1].tolist()
    all_z = gt_z[:frame+1] + pred_z[:frame+1].tolist()
    if all_x:
        pad = 0.1
        x_min, x_max = min(all_x), max(all_x)
        z_min, z_max = min(all_z), max(all_z)
        ax_traj.set_xlim(x_min - pad, x_max + pad)
        ax_traj.set_ylim(z_min - pad, z_max + pad)
    
    return video_display, gt_line, pred_line

# Create animation
print("Creating animation...")
ani = FuncAnimation(
    fig=fig,
    func=update,
    frames=n_frames,
    interval=100,
    blit=False
)

# Save as GIF
output_path = "visual_odometry_7dof.gif"
print(f"Saving animation to {output_path}...")
ani.save(
    output_path,
    writer='pillow',
    dpi=100,
    bitrate=1800,
    progress_callback=lambda i, n: print(f"Processing frame {i+1}/{n}", end='\r')
)

print("\nAnimation saved successfully!")
```

```python
import numpy as np

def read_kitti_gt_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()

    transformation_matrices = []
    for line in lines:
        values = [float(value) for value in line.split()]
        transformation_matrix = np.array(values).reshape(3, 4)
        transformation_matrices.append(transformation_matrix)

    return np.array(transformation_matrices)

gt = read_kitti_gt_file('/Users/epheriami/Downloads/4B/Code/SYDE 673/P/data/poses/03.txt')
est = read_kitti_gt_file('/Users/epheriami/Downloads/4B/Code/SYDE 673/P/tsformer/checkpoints/Exp3/checkpoint_model3_exp20/pred_poses/03.txt')
```

```python
import numpy as np

def calculate_displacement(gt, est):
    """
    Calculates the max and RMS positional and rotational displacement between two sets of transformation matrices.

    Args:
        gt (np.ndarray): Ground truth transformation matrices.
        est (np.ndarray): Estimated transformation matrices.

    Returns:
        dict: A dictionary containing the max and RMS positional and rotational displacement.
    """
    # Ensure both arrays have the same shape
    assert gt.shape == est.shape, "Both arrays must have the same shape"

    # Calculate the positional displacement
    positional_displacement = np.linalg.norm(gt[:, :3, 3] - est[:, :3, 3], axis=1)

    # Calculate the rotational displacement
    rotational_displacement = np.zeros((gt.shape[0],))
    for i in range(gt.shape[0]):
        gt_rot = gt[i, :3, :3]
        est_rot = est[i, :3, :3]
        rot_diff = np.dot(gt_rot.T, est_rot)
        axis_angle = np.array([rot_diff[2, 1], rot_diff[0, 2], rot_diff[1, 0]])
        rotational_displacement[i] = np.linalg.norm(axis_angle)

    # Calculate the max and RMS displacement
    max_positional_displacement = np.max(positional_displacement)
    rms_positional_displacement = np.sqrt(np.mean(positional_displacement ** 2))
    max_rotational_displacement = np.max(rotational_displacement)
    rms_rotational_displacement = np.sqrt(np.mean(rotational_displacement ** 2))

    return {
        "max_positional_displacement": round(max_positional_displacement, 2),
        "rms_positional_displacement": round(rms_positional_displacement, 2),
        "max_rotational_displacement": round(max_rotational_displacement, 2),
        "rms_rotational_displacement": round(rms_rotational_displacement, 2),
    }

displacement = calculate_displacement(gt[4:], est)
print(f"Max Positional Displacement: {displacement['max_positional_displacement']} m.")
print(f"RMS Positional Displacement: {displacement['rms_positional_displacement']} m.")
print(f"Max Rotational Displacement: {displacement['max_rotational_displacement']}°")
print(f"RMS Rotational Displacement: {displacement['rms_rotational_displacement']}°")
```

```python
import os
from glob import glob
from PIL import Image
from tqdm import tqdm

""" Convert png to jpg images for KITTI dataset

Data structure:
|---dataset
    |---sequences
        |---00
            |---image_0
                |---000000.png
                |---000001.png
                |---...
            |---image_1
                |...
            |---image_2
                |---...
            |---image_3
                |---...
        |---01
        |---...
"""

dir_path = "/Users/epheriami/Downloads/4B/Code/SYDE 673/P/dataset"

dataset_dir = f"{dir_path}/sequences"
new_root_dir = f"{dir_path}/sequences_jpg"
image_nb = 0  # number of image folder (image_0, image_1, ... image_3)

# create new directory
if not os.path.exists(new_root_dir):
    os.makedirs(new_root_dir)

for seq_nb in range(22): # list all sequences ["00", ... "21"]
    # sequence as 2-digit string
    seq = "{:02d}".format(seq_nb)
    
    # create seq in save directory
    seq_dir = os.path.join(new_root_dir, seq)
    if not os.path.exists(seq_dir):
        os.makedirs(seq_dir)
        
    # create image_nb directory
    img_nb_dir = os.path.join(seq_dir, "image_{}".format(image_nb))
    if not os.path.exists(img_nb_dir):
        os.makedirs(img_nb_dir)
    
    images_list = glob(os.path.join(dataset_dir, seq, "image_{}".format(image_nb), "*"))  # paths to png images in seq
    
    for i in tqdm(range(len(images_list)), desc="Sequence {}: ".format(seq)):
        img_path = images_list[i]
        img_name = os.path.basename(img_path)
        
        # read png image and convert to jpg using PIL
        img = Image.open(img_path)
        save_dir = os.path.join(img_nb_dir, img_name.replace(".png", ".jpg"))
        img.save(save_dir)
```

```python
import pickle
import os
import numpy as np
from torch import nn

def params_count(model, ignore_bn=False):
    """
    Compute the number of parameters.
    Args:
        model (model): model to count the number of parameters.
    """
    if not ignore_bn:
        return np.sum([p.numel() for p in model.parameters()]).item()
    else:
        count = 0
        for m in model.modules():
            if not isinstance(m, nn.BatchNorm3d):
                for p in m.parameters(recurse=False):
                    count += p.numel()
    return count


with open(os.path.join("/Users/epheriami/Downloads/4B/Code/SYDE 673/P/tsformer/checkpoints/Exp3/args.pkl"), 'rb') as f:
    args = pickle.load(f)

f.close()
model_params = args["model_params"]

model = VisionTransformer(img_size=model_params["image_size"],
                          num_classes=model_params["num_classes"],
                          patch_size=model_params["patch_size"],
                          embed_dim=model_params["dim"],
                          depth=model_params["depth"],
                          num_heads=model_params["heads"],
                          mlp_ratio=4,
                          qkv_bias=True,
                          norm_layer=partial(nn.LayerNorm, eps=1e-6),
                          drop_rate=0.,
                          attn_drop_rate=0.,
                          drop_path_rate=0.1,
                          num_frames=model_params["num_frames"],
                          attention_type=model_params["attention_type"])


params_count(model)
```
